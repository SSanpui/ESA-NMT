```python
# Colab-ready: IndicTrans2 inference + metrics (BLEU, METEOR, ROUGE-L, CHRF,
# emotion-preservation-accuracy, semantic-accuracy)
# Tested conceptually with instructions from IndicTrans2 model card / colab example.
# Requires GPU for reasonable speed.

# 1) Install dependencies (run once)
!pip install -q transformers==4.35.0 accelerate sentence-transformers IndicTransToolkit evaluate sacrebleu rouge_score nltk

# (some installs print lots of output; that's expected)
# If IndicTransToolkit fails with pip, the repo can be cloned and installed editable as shown in the IndicTrans2 repo.

# 2) Imports
import os
import torch
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer, pipeline
from IndicTransToolkit.processor import IndicProcessor
from sentence_transformers import SentenceTransformer, util
import evaluate
import pandas as pd
import numpy as np
from tqdm.auto import tqdm
import nltk
nltk.download('punkt')  # for METEOR tokenizer

# 3) Configuration - change as needed
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
model_name = "ai4bharat/indictrans2-indic-indic-dist-320M"   # distilled Indic-Indic model (smaller / feasible)
# model_name = "ai4bharat/indictrans2-indic-indic-1B"        # larger (if you have more GPU)
batch_size = 8
max_length = 256
num_beams = 5

# 4) Language codes used by IndicTrans2 (examples)
LANG_CODE = {
    "ben": "ben_Beng",   # Bengali
    "hin": "hin_Deva",   # Hindi
    "tel": "tel_Telu",   # Telugu
    "eng": "eng_Latn"    # English (for emotion-checking)
}

# 5) Load model + tokenizer + IndicProcessor
print("Loading IndicTrans2 model (this may take a minute)...")
tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name, trust_remote_code=True).to(DEVICE)
ip = IndicProcessor(inference=True)   # preprocess/postprocess helpers

# 6) Load metric backends (HuggingFace evaluate) - BLEU, METEOR, rouge, chrf
bleu_metric = evaluate.load("sacrebleu")
meteor_metric = evaluate.load("meteor")
rouge_metric = evaluate.load("rouge")
chrf_metric = evaluate.load("chrf")

# 7) Sentence-transformer for semantic similarity (multilingual)
sbert = SentenceTransformer("sentence-transformers/paraphrase-multilingual-mpnet-base-v2")

# 8) English emotion classifier (we use an English emotion model; we translate texts to English first)
#    You may replace the model with another English emotion classifier you prefer.
emotion_clf = pipeline("text-classification", model="j-hartmann/emotion-english-distilroberta-base", return_all_scores=False, device=0 if DEVICE=="cuda" else -1)

# 9) Helper: translate a list of sentences from src_lang to tgt_lang using IndicTrans2
def translate_batch(sentences, src_lang_code, tgt_lang_code):
    """
    sentences: list[str] in source language (native script)
    src_lang_code, tgt_lang_code: IndicTrans2 language codes, e.g. 'ben_Beng', 'hin_Deva'
    returns: list[str] translated sentences (postprocessed)
    """
    # preprocessing using IndicProcessor: returns list of preprocessed strings
    batch = ip.preprocess_batch(sentences, src_lang=src_lang_code, tgt_lang=tgt_lang_code)
    # tokenize
    inputs = tokenizer(batch, truncation=True, padding="longest", return_tensors="pt").to(DEVICE)
    with torch.no_grad():
        gen = model.generate(
            **inputs,
            max_length=max_length,
            num_beams=num_beams,
            pad_token_id=tokenizer.pad_token_id if tokenizer.pad_token_id is not None else model.config.pad_token_id,
            eos_token_id=tokenizer.eos_token_id if tokenizer.eos_token_id is not None else model.config.eos_token_id
        )
    decoded = tokenizer.batch_decode(gen, skip_special_tokens=True, clean_up_tokenization_spaces=True)
    translations = ip.postprocess_batch(decoded, lang=tgt_lang_code)
    return translations

# 10) Metric wrapper to evaluate lists of hypotheses vs references
def compute_metrics(hyps, refs):
    # sacrebleu expects detokenized strings; evaluate's sacrebleu uses references as list-of-lists
    bleu = bleu_metric.compute(predictions=hyps, references=[[r] for r in refs])
    meteor = meteor_metric.compute(predictions=hyps, references=refs)
    rouge = rouge_metric.compute(predictions=hyps, references=refs)
    chrf = chrf_metric.compute(predictions=hyps, references=refs)
    return {"bleu": bleu["score"], "meteor": meteor["meteor"], "rougeL": rouge["rougeL"], "chrf": chrf["score"]}

# 11) Emotion preservation accuracy
def emotion_preservation_accuracy(src_texts, hyp_texts, src_lang, tgt_lang):
    """
    Strategy:
      - Translate source texts to English (src_lang -> eng_Latn)
      - Translate hypothesis texts to English (tgt_lang -> eng_Latn)
      - Use English emotion classifier to predict emotion labels
      - Accuracy = fraction where label(src_en) == label(hyp_en)
    """
    # translate batches to English
    src_en = translate_in_batches(src_texts, src_lang, LANG_CODE["eng"])
    hyp_en = translate_in_batches(hyp_texts, tgt_lang, LANG_CODE["eng"])
    # predict emotion labels
    src_labels = [emotion_clf(x)[0]["label"] for x in src_en]
    hyp_labels = [emotion_clf(x)[0]["label"] for x in hyp_en]
    matches = [1 if a==b else 0 for a,b in zip(src_labels, hyp_labels)]
    return {"accuracy": float(np.mean(matches)), "n": len(matches), "src_labels": src_labels[:5], "hyp_labels": hyp_labels[:5]}

# small helper to translate in batches (robust for large lists)
def translate_in_batches(sentences, src_lang_code, tgt_lang_code, bsz=batch_size):
    outs = []
    for i in range(0, len(sentences), bsz):
        batch = sentences[i:i+bsz]
        outs += translate_batch(batch, src_lang_code, tgt_lang_code)
    return outs

# 12) Semantic accuracy using multilingual SBERT embeddings
def semantic_accuracy(hyps, refs, threshold=0.65):
    """
    Compute cosine similarity between hypothesis and reference embeddings.
    Returns fraction of pairs with similarity >= threshold.
    """
    emb_h = sbert.encode(hyps, convert_to_tensor=True, show_progress_bar=False)
    emb_r = sbert.encode(refs, convert_to_tensor=True, show_progress_bar=False)
    sims = util.cos_sim(emb_h, emb_r).diagonal().cpu().numpy()  # pairwise diagonal
    acc = float((sims >= threshold).astype(float).mean())
    return {"semantic_accuracy": acc, "mean_similarity": float(sims.mean()), "similarities_sample": sims[:10].tolist()}

# 13) Example: load your dataset (CSV) and run
# Expect CSV files with columns: "src" (source sentence), "ref" (reference translation),
# and an optional "lang_pair" or you can run three separate files per pair.
#
# For convenience, I'll show a sample pipeline for a single CSV and a specified language pair.

# -- CONFIGURE FILE PATH + language pair you want to evaluate --
# Replace this with your file path in Colab (you can upload file via left pane)
# Example CSV columns: src, ref
csv_path = "/content/ESA-NMT/BHT25_All.csv"   # please change
# set your language pair:
SRC = LANG_CODE["ben"]   # 'ben_Beng'
TGT = LANG_CODE["hin"]   # 'hin_Deva'

# 14) Load CSV and run inference (example)
df = pd.read_csv(csv_path)
# The dataframe has columns 'bn', 'hi', 'te'. Use 'bn' for Bengali source and 'hi' for Hindi reference.
src_texts = df['bn'].astype(str).tolist()
refs = df['hi'].astype(str).tolist()

# (optionally limit to N for quick runs)
N = min(len(src_texts), 200)   # you can increase / decrease
src_texts = src_texts[:N]
refs = refs[:N]

print(f"Translating {len(src_texts)} sentences from {SRC} -> {TGT} ...")
hyps = translate_in_batches(src_texts, SRC, TGT)

# 15) Compute conventional metrics
metrics = compute_metrics(hyps, refs)
print("Conventional metrics (BLEU/METEOR/ROUGE-L/ChrF):")
print(metrics)

# 16) Compute emotion-preservation accuracy (source->eng, hypothesis->eng -> emotion classifier)
print("Computing emotion preservation accuracy (via English emotion classifier)...")
emo_stats = emotion_preservation_accuracy(src_texts, hyps, SRC, TGT)
print(emo_stats)

# 17) Compute semantic accuracy (SBERT similarity)
print("Computing semantic accuracy (SBERT cosine similarity)... வளர்ச்சி)")
sem_stats = semantic_accuracy(hyps, refs, threshold=0.65)
print(sem_stats)

# 18) Save results back to dataframe and CSV
df_out = pd.DataFrame({
    "src": src_texts,
    "ref": refs,
    "hyp": hyps
})
df_out["sbert_sim"] = sbert.encode(hyps, convert_to_tensor=False).tolist()  # optional: heavy for large N; you can compute per row
# (Above line may not be ideal for large N; instead compute similarity batchwise if needed.)

df_out.to_csv("/content/translation_results_sample.csv", index=False)
print("Saved /content/translation_results_sample.csv")
```

```python
#Example.py from huggingface-AI4bharat model
import sys
import torch
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer, BitsAndBytesConfig
from transformers.utils import is_flash_attn_2_available, is_flash_attn_greater_or_equal_2_10

# Install IndicTransToolkit if not already present
try:
    from IndicTransToolkit.processor import IndicProcessor
except ImportError:
    !pip install -q IndicTransToolkit
    from IndicTransToolkit.processor import IndicProcessor

from mosestokenizer import MosesSentenceSplitter
from nltk import sent_tokenize
from indicnlp.tokenize.sentence_tokenize import sentence_split, DELIM_PAT_NO_DANDA


en_indic_ckpt_dir = "ai4bharat/indictrans2-en-indic-1B"  # ai4bharat/indictrans2-en-indic-dist-200M
indic_en_ckpt_dir = "ai4bharat/indictrans2-indic-en-1B"  # ai4bharat/indictrans2-indic-en-dist-200M
indic_indic_ckpt_dir = (
    "ai4bharat/indictrans2-indic-indic-dist-320M"  # ai4bharat/indictrans2-indic-indic-dist-320M
)
BATCH_SIZE = 4
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

if len(sys.argv) > 1:
    quantization = sys.argv[1]
    attn_implementation = sys.argv[2]
else:
    quantization = ""
    attn_implementation = "eager"


# FLORES language code mapping to 2 letter ISO language code for compatibility
# with Indic NLP Library (https://github.com/anoopkunchukuttan/indic_nlp_library)
flores_codes = {
    "asm_Beng": "as",
    "awa_Deva": "hi",
    "ben_Beng": "bn",
    "bho_Deva": "hi",
    "brx_Deva": "hi",
    "doi_Deva": "hi",
    "eng_Latn": "en",
    "gom_Deva": "kK",
    "guj_Gujr": "gu",
    "hin_Deva": "hi",
    "hne_Deva": "hi",
    "kan_Knda": "kn",
    "kas_Arab": "ur",
    "kas_Deva": "hi",
    "kha_Latn": "en",
    "lus_Latn": "en",
    "mag_Deva": "hi",
    "mai_Deva": "hi",
    "mal_Mlym": "ml",
    "mar_Deva": "mr",
    "mni_Beng": "bn",
    "mni_Mtei": "hi",
    "npi_Deva": "ne",
    "ory_Orya": "or",
    "pan_Guru": "pa",
    "san_Deva": "hi",
    "sat_Olck": "or",
    "snd_Arab": "ur",
    "snd_Deva": "hi",
    "tam_Taml": "ta",
    "tel_Telu": "te",
    "urd_Arab": "ur",
}


def split_sentences(input_text, lang):
    if lang == "eng_Latn":
        input_sentences = sent_tokenize(input_text)
        with MosesSentenceSplitter(flores_codes[lang]) as splitter:
            sents_moses = splitter([input_text])
        sents_nltk = sent_tokenize(input_text)
        if len(sents_nltk) < len(sents_moses):
            input_sentences = sents_nltk
        else:
            input_sentences = sents_moses
        input_sentences = [sent.replace("\xad", "") for sent in input_sentences]
    else:
        input_sentences = sentence_split(
            input_text, lang=flores_codes[lang], delim_pat=DELIM_PAT_NO_DANDA
        )
    return input_sentences


def initialize_model_and_tokenizer(ckpt_dir, quantization, attn_implementation):
    if quantization == "4-bit":
        qconfig = BitsAndBytesConfig(
            load_in_4bit=True,
            bnb_4bit_use_double_quant=True,
            bnb_4bit_compute_dtype=torch.bfloat16,
        )
    elif quantization == "8-bit":
        qconfig = BitsAndBytesConfig(
            load_in_8bit=True,
            bnb_8bit_use_double_quant=True,
            bnb_8bit_compute_dtype=torch.bfloat16,
        )
    else:
        qconfig = None

    if attn_implementation == "flash_attention_2":
        if is_flash_attn_2_available() and is_flash_attn_greater_or_equal_2_10():
            attn_implementation = "flash_attention_2"
        else:
            attn_implementation = "eager"

    tokenizer = AutoTokenizer.from_pretrained(ckpt_dir, trust_remote_code=True)
    model = AutoModelForSeq2SeqLM.from_pretrained(
        ckpt_dir,
        trust_remote_code=True,
        attn_implementation=attn_implementation,
        low_cpu_mem_usage=True,
        quantization_config=qconfig,
    )

    if qconfig == None:
        model = model.to(DEVICE)
        model.half()

    model.eval()

    return tokenizer, model


def batch_translate(input_sentences, src_lang, tgt_lang, model, tokenizer, ip):
    translations = []
    for i in range(0, len(input_sentences), BATCH_SIZE):
        batch = input_sentences[i : i + BATCH_SIZE]

        # Preprocess the batch and extract entity mappings
        batch = ip.preprocess_batch(batch, src_lang=src_lang, tgt_lang=tgt_lang)

        # Tokenize the batch and generate input encodings
        inputs = tokenizer(
            batch,
            truncation=True,
            padding="longest",
            return_tensors="pt",
            return_attention_mask=True,
        ).to(DEVICE)

        # Generate translations using the model
        with torch.no_grad():
            generated_tokens = model.generate(
                **inputs,
                use_cache=True,
                min_length=0,
                max_length=256,
                num_beams=5,
                num_return_sequences=1,
            )

        # Decode the generated tokens into text
        generated_tokens = tokenizer.batch_decode(
            generated_tokens,
            skip_special_tokens=True,
            clean_up_tokenization_spaces=True,
        )

        # Postprocess the translations, including entity replacement
        translations += ip.postprocess_batch(generated_tokens, lang=tgt_lang)

        del inputs
        torch.cuda.empty_cache()

    return translations


def translate_paragraph(input_text, src_lang, tgt_lang, model, tokenizer, ip):
    input_sentences = split_sentences(input_text, src_lang)
    translated_text = batch_translate(input_sentences, src_lang, tgt_lang, model, tokenizer, ip)
    return " ".join(translated_text)


ip = IndicProcessor(inference=True)

en_indic_tokenizer, en_indic_model = initialize_model_and_tokenizer(
    en_indic_ckpt_dir, quantization, attn_implementation
)

indic_en_tokenizer, indic_en_model = initialize_model_and_tokenizer(
    indic_en_ckpt_dir, quantization, attn_implementation
)

indic_indic_tokenizer, indic_indic_model = initialize_model_and_tokenizer(
    indic_indic_ckpt_dir, quantization, attn_implementation
)

# ---------------------------------------------------------------------------
#                              Hindi to English
# ---------------------------------------------------------------------------
hi_sents = [
    "जब मैं छोटा था, मैं हर रोज़ पार्क जाता था।",
    "उसके पास बहुत सारी पुरानी किताबें हैं, जिन्हें उसने अपने दादा-परदादा से विरासत में पाया।",
    "मुझे समझ में नहीं आ रहा कि मैं अपनी समस्या का समाधान कैसे ढूंढूं।",
    "वह बहुत मेहनती और समझदार है, इसलिए उसे सभी अच्छे मार्क्स मिले।",
    "हमने पिछले सप्ताह एक नई फिल्म देखी जो कि बहुत प्रेरणादायक थी।",
    "अगर तुम मुझे उस समय पास मिलते, तो हम बाहर खाना खाने चलते।",
    "वह अपनी दीदी के साथ बाजार गयी थी ताकि वह नई साड़ी खरीद सके।",
    "राज ने मुझसे कहा कि वह अगले महीने अपनी नानी के घर जा रहा है।",
    "सभी बच्चे पार्टी में मज़ा कर रहे थे और खूब सारी मिठाइयाँ खा रहे थे।",
    "मेरे मित्र ने मुझे उसके जन्मदिन की पार्टी में बुलाया है, और मैं उसे एक तोहफा दूंगा।",
]
src_lang, tgt_lang = "hin_Deva", "eng_Latn"
en_translations = batch_translate(
    hi_sents, src_lang, tgt_lang, indic_en_model, indic_en_tokenizer, ip
)

print(f"\n{src_lang} - {tgt_lang}")
for input_sentence, translation in zip(hi_sents, en_translations):
    print(f"{src_lang}: {input_sentence}")
    print(f"{tgt_lang}: {translation}")


# ---------------------------------------------------------------------------
#                              English to Hindi
# ---------------------------------------------------------------------------
en_sents = [
    "When I was young, I used to go to the park every day.",
    "He has many old books, which he inherited from his ancestors.",
    "I can't figure out how to solve my problem.",
    "She is very hardworking and intelligent, which is why she got all the good marks.",
    "We watched a new movie last week, which was very inspiring.",
    "If you had met me at that time, we would have gone out to eat.",
    "She went to the market with her sister to buy a new sari.",
    "Raj told me that he is going to his grandmother's house next month.",
    "All the kids were having fun at the party and were eating lots of sweets.",
    "My friend has invited me to his birthday party, and I will give him a gift.",
]
src_lang, tgt_lang = "eng_Latn", "hin_Deva"
hi_translations = batch_translate(
    en_sents, src_lang, tgt_lang, en_indic_model, en_indic_tokenizer, ip
)

print(f"\n{src_lang} - {tgt_lang}")
for input_sentence, translation in zip(en_sents, hi_translations):
    print(f"{src_lang}: {input_sentence}")
    print(f"{tgt_lang}: {translation}")


# ---------------------------------------------------------------------------
#                              Hindi to Marathi
# ---------------------------------------------------------------------------
hi_sents = [
    "जब मैं छोटा था, मैं हर रोज़ पार्क जाता था।",
    "उसके पास बहुत सारी पुरानी किताबें हैं, जिन्हें उसने अपने दादा-परदादा से विरासत में पाया।",
    "मुझे समझ में नहीं आ रहा कि मैं अपनी समस्या का समाधान कैसे ढूंढूं।",
    "वह बहुत मेहनती और समझदार है, इसलिए उसे सभी अच्छे मार्क्स मिले।",
    "हमने पिछले सप्ताह एक नई फिल्म देखी जो कि बहुत प्रेरणादायक थी।",
    "अगर तुम मुझे उस समय पास मिलते, तो हम बाहर खाना खाने चलते।",
    "वह अपनी दीदी के साथ बाजार गयी थी ताकि वह नई साड़ी खरीद सके।",
    "राज ने मुझसे कहा कि वह अगले महीने अपनी नानी के घर जा रहा है।",
    "सभी बच्चे पार्टी में मज़ा कर रहे थे और खूब सारी मिठाइयाँ खा रहे थे।",
    "मेरे मित्र ने मुझे उसके जन्मदिन की पार्टी में बुलाया है, और मैं उसे एक तोहफा दूंगा।",
]
src_lang, tgt_lang = "hin_Deva", "mar_Deva"
mr_translations = batch_translate(
    hi_sents, src_lang, tgt_lang, indic_indic_model, indic_indic_tokenizer, ip
)

print(f"\n{src_lang} - {tgt_lang}")
for input_sentence, translation in zip(hi_sents, mr_translations):
    print(f"{src_lang}: {input_sentence}")
    print(f"{tgt_lang}: {translation}")


# ---------------------------------------------------------------------------
#                             Paragraph translation
# ---------------------------------------------------------------------------
src_lang, tgt_lang = "hin_Deva", "eng_Latn"
hi_text = "यहाँ एक पाराग्राफ है जो हिंदी में लिखा गया है। हिंदी एक सुंदर भाषा है और भारत की राष्ट्रीय भाषा है। इसका विकास विभिन्न कालों में हुआ है और यह विशेषतः भारतीय उपमहाद्वीप में बोली जाती है। हिंदी भाषा का साहित्य, संस्कृति और इतिहास भी बहुत गर्वनीय है."
en_translated_text = translate_paragraph(
    hi_text, src_lang, tgt_lang, indic_en_model, indic_en_tokenizer, ip
)
print(f"{src_lang}: {hi_text}")
print(f"{tgt_lang}: {en_translated_text}")
```

# **mBART-50 Baseline Evaluation**

```python
#!/usr/bin/env python3
"""
mBART-50 Baseline Evaluation
NO TRAINING REQUIRED - Just evaluate pre-trained mBART-50 on your BHT25 dataset

Evaluates 3 language pairs:
- Bengali → Hindi (bn-hi)
- Bengali → Telugu (bn-te)
- Hindi → Telugu (hi-te)

Usage:
    python evaluate_mbart50.py
"""

import os
import torch
import pandas as pd
import numpy as np
from tqdm import tqdm
import json
from datetime import datetime
import gc

print("="*80)
print("mBART-50 BASELINE EVALUATION")
print("Pre-trained model - NO TRAINING NEEDED")
print("="*80 + "\n")

# ============================================================================
# CONFIGURATION
# ============================================================================

# UPDATE THIS PATH to your BHT25 dataset
DATA_FILE = '/content/ESA-NMT/BHT25_All.csv'
# OR if you have original BHT25:
# DATA_FILE = '/kaggle/input/bht25/BHT25.csv'

OUTPUT_DIR = './outputs'
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'

# mBART-50 language codes
MBART_LANG_CODES = {
    'bn-hi': ('bn_IN', 'hi_IN', 'bn', 'hi'),  # Bengali to Hindi
    'bn-te': ('bn_IN', 'te_IN', 'bn', 'te'),  # Bengali to Telugu
    'hi-te': ('hi_IN', 'te_IN', 'hi', 'te')   # Hindi to Telugu
}

# Number of test samples (set None for all)
TEST_LIMIT = 2000  # Adjust based on time: 2000 samples ≈ 30-40 minutes

print(f"Configuration:")
print(f"  Device: {DEVICE}")
print(f"  Dataset: {DATA_FILE}")
print(f"  Test samples: {TEST_LIMIT if TEST_LIMIT else 'All'}")
print(f"  Output: {OUTPUT_DIR}\n")

os.makedirs(OUTPUT_DIR, exist_ok=True)

# ============================================================================
# LOAD DATASET
# ============================================================================

def load_bht25_data(csv_path, pair):
    """
    Load BHT25 dataset for a specific language pair

    Args:
        csv_path: Path to BHT25 CSV file
        pair: Language pair ('bn-hi', 'bn-te', 'hi-te')

    Returns:
        sources, references (test split only)
    """
    print(f"\n{'='*70}")
    print(f"Loading dataset for {pair}")
    print(f"{'='*70}\n")

    df = pd.read_csv(csv_path)
    print(f"Total samples: {len(df)}")
    print(f"Columns: {df.columns.tolist()}\n")

    # Get column names
    _, _, src_col, tgt_col = MBART_LANG_CODES[pair]

    # Check columns exist
    if src_col not in df.columns or tgt_col not in df.columns:
        print(f"❌ ERROR: Columns '{src_col}' or '{tgt_col}' not found!")
        print(f"   Available: {df.columns.tolist()}")
        return None, None

    # Split dataset: 70% train, 15% val, 15% test (same as your training)
    train_size = int(len(df) * 0.70)
    val_size = int(len(df) * 0.15)
    test_start = train_size + val_size

    # Get test split
    test_df = df[test_start:]

    # Apply limit if specified
    if TEST_LIMIT and len(test_df) > TEST_LIMIT:
        test_df = test_df.head(TEST_LIMIT)

    # Extract source and target texts
    sources = test_df[src_col].astype(str).tolist()
    references = test_df[tgt_col].astype(str).tolist()

    print(f"✅ Loaded test data:")
    print(f"   Test samples: {len(sources)}")
    print(f"   Source column: {src_col}")
    print(f"   Target column: {tgt_col}\n")

    # Show sample
    print("Sample (first entry):")
    print(f"  SRC: {sources[0][:100]}...")
    print(f"  REF: {references[0][:100]}...\n")

    return sources, references

# ============================================================================
# mBART-50 TRANSLATOR
# ============================================================================

class mBART50Translator:
    """
    Pre-trained mBART-50 translator for Indian languages
    NO TRAINING REQUIRED
    """

    def __init__(self, src_lang, tgt_lang):
        """
        Initialize mBART-50 translator

        Args:
            src_lang: Source language code (bn_IN, hi_IN, te_IN)
            tgt_lang: Target language code (bn_IN, hi_IN, te_IN)
        """
        self.src_lang = src_lang
        self.tgt_lang = tgt_lang
        self.device = DEVICE

        print(f"Loading mBART-50 model...")
        print(f"  Translation: {src_lang} → {tgt_lang}")

        from transformers import MBartForConditionalGeneration, MBart50TokenizerFast

        # Load pre-trained mBART-50
        model_name = "facebook/mbart-large-50-many-to-many-mmt"

        self.tokenizer = MBart50TokenizerFast.from_pretrained(model_name)
        self.model = MBartForConditionalGeneration.from_pretrained(model_name)

        # Move to GPU
        self.model = self.model.to(self.device)
        self.model.eval()

        # Set source language
        self.tokenizer.src_lang = src_lang

        print(f"✅ mBART-50 loaded successfully\n")

    def translate_batch(self, texts, batch_size=8):
        """
        Translate a batch of texts

        Args:
            texts: List of source texts
            batch_size: Number of texts to translate at once

        Returns:
            List of translated texts
        """
        translations = []

        print(f"Translating {len(texts)} samples...")

        with torch.no_grad():
            for i in tqdm(range(0, len(texts), batch_size), desc="mBART-50"):
                batch_texts = texts[i:i+batch_size]

                try:
                    # Tokenize
                    encoded = self.tokenizer(
                        batch_texts,
                        return_tensors="pt",
                        padding=True,
                        truncation=True,
                        max_length=256
                    ).to(self.device)

                    # Generate translations
                    generated = self.model.generate(
                        **encoded,
                        forced_bos_token_id=self.tokenizer.lang_code_to_id[self.tgt_lang],
                        max_length=256,
                        num_beams=5,
                        length_penalty=1.0,
                        early_stopping=True
                    )

                    # Decode
                    batch_translations = self.tokenizer.batch_decode(
                        generated,
                        skip_special_tokens=True
                    )

                    translations.extend(batch_translations)

                except Exception as e:
                    print(f"\n⚠️ Error in batch {i//batch_size}: {e}")
                    # Fallback: add empty translations
                    translations.extend([''] * len(batch_texts))

                # Memory cleanup every 50 batches
                if (i // batch_size) % 50 == 0:
                    torch.cuda.empty_cache()

        print(f"✅ Translation completed: {len(translations)} samples\n")

        return translations

    def cleanup(self):
        """Free GPU memory"""
        del self.model
        del self.tokenizer
        torch.cuda.empty_cache()
        gc.collect()

# ============================================================================
# EVALUATION METRICS
# ============================================================================

def calculate_metrics(predictions, references):
    """
    Calculate all evaluation metrics

    Args:
        predictions: List of translated texts
        references: List of reference translations

    Returns:
        Dictionary of metrics
    """
    print("Computing metrics...")

    # Install required packages
    try:
        import sacrebleu
        from rouge_score import rouge_scorer
        import nltk
        nltk.download('wordnet', quiet=True)
        nltk.download('omw-1.4', quiet=True)
        nltk.download('punkt', quiet=True)
    except:
        print("⚠️ Installing required packages...")
        import subprocess
        subprocess.run(['pip', 'install', '-q', 'sacrebleu', 'rouge-score', 'nltk'])

    metrics = {}

    # 1. BLEU Score
    print("  - BLEU...")
    from sacrebleu import corpus_bleu
    bleu = corpus_bleu(predictions, [references])
    metrics['BLEU'] = float(round(bleu.score, 2)) # Explicitly cast to float

    # 2. chrF++ Score
    print("  - chrF++...")
    from sacrebleu.metrics import CHRF
    chrf = CHRF()
    chrf_score = chrf.corpus_score(predictions, [references])
    metrics['chrF'] = float(round(chrf_score.score, 2)) # Explicitly cast to float

    # 3. METEOR Score
    print("  - METEOR...")
    try:
        from nltk.translate.meteor_score import meteor_score as calc_meteor
        import nltk

        meteor_scores = []
        for pred, ref in zip(predictions, references):
            try:
                # Tokenize
                pred_tokens = nltk.word_tokenize(pred) if pred else ['']
                ref_tokens = nltk.word_tokenize(ref) if ref else ['']
                score = calc_meteor([ref_tokens], pred_tokens)
                meteor_scores.append(score)
            except:
                meteor_scores.append(0.0)

        metrics['METEOR'] = float(round(np.mean(meteor_scores) * 100, 2)) # Explicitly cast to float
    except Exception as e:
        print(f"    ⚠️ METEOR calculation failed: {e}")
        metrics['METEOR'] = 0.0

    # 4. ROUGE-L Score
    print("  - ROUGE-L...")
    try:
        rouge = rouge_scorer.RougeScorer(['rougeL'], use_stemmer=False)
        rouge_scores = []

        for pred, ref in zip(predictions, references):
            if pred and ref:
                score = rouge.score(ref, pred)
                rouge_scores.append(score['rougeL'].fmeasure)
            else:
                rouge_scores.append(0.0)

        metrics['ROUGE-L'] = float(round(np.mean(rouge_scores), 4)) # Explicitly cast to float
    except Exception as e:
        print(f"    ⚠️ ROUGE-L calculation failed: {e}")
        metrics['ROUGE-L'] = 0.0

    # 5. Semantic Similarity (using LaBSE)
    print("  - Semantic Similarity...")
    try:
        from sentence_transformers import SentenceTransformer

        semantic_model = SentenceTransformer('sentence-transformers/LaBSE').to(DEVICE)

        # Compute in batches
        batch_size = 32
        all_cos_sims = []

        for i in range(0, len(predictions), batch_size):
            batch_src = references[i:i+batch_size]  # Use references as "source"
            batch_pred = predictions[i:i+batch_size]

            # Filter out empty strings
            valid_pairs = [(s, p) for s, p in zip(batch_src, batch_pred) if s and p]
            if not valid_pairs:
                continue

            batch_src_valid, batch_pred_valid = zip(*valid_pairs)

            src_emb = semantic_model.encode(list(batch_src_valid), show_progress_bar=False)
            pred_emb = semantic_model.encode(list(batch_pred_valid), show_progress_bar=False)

            # Cosine similarity
            cos_sims = np.sum(src_emb * pred_emb, axis=1) / (
                np.linalg.norm(src_emb, axis=1) * np.linalg.norm(pred_emb, axis=1) + 1e-8
            )
            all_cos_sims.extend(cos_sims)

        metrics['Semantic'] = float(round(np.mean(all_cos_sims), 4)) # Explicitly cast to float

        del semantic_model
        torch.cuda.empty_cache()

    except Exception as e:
        print(f"    ⚠️ Semantic similarity calculation failed: {e}")
        metrics['Semantic'] = 0.0

    print("✅ Metrics computed\n")

    return metrics

# ============================================================================
# EVALUATION PIPELINE
# ============================================================================

def evaluate_mbart50(pair, data_path):
    """
    Evaluate mBART-50 on one language pair

    Args:
        pair: Language pair ('bn-hi', 'bn-te', 'hi-te')
        data_path: Path to BHT25 CSV file

    Returns:
        Dictionary with results
    """
    print(f"\n{'='*80}")
    print(f"EVALUATING: {pair.upper()}")
    print(f"{'='*80}\n")

    # Get language codes
    src_mbart, tgt_mbart, src_col, tgt_col = MBART_LANG_CODES[pair]

    # Load data
    sources, references = load_bht25_data(data_path, pair)

    if sources is None:
        print(f"❌ Failed to load data for {pair}\n")
        return None

    # Initialize translator
    translator = mBART50Translator(src_mbart, tgt_mbart)

    # Translate
    predictions = translator.translate_batch(sources, batch_size=8)

    # Show sample translations
    print("Sample translations (first 3):")
    for i in range(min(3, len(predictions))):
        print(f"\n{i+1}.")
        print(f"  SRC:   {sources[i][:80]}...")
        print(f"  REF:   {references[i][:80]}...")
        print(f"  mBART: {predictions[i][:80]}...")
    print()

    # Calculate metrics
    metrics = calculate_metrics(predictions, references)

    # Print results
    print(f"\n{'='*70}")
    print(f"mBART-50 RESULTS: {pair}")
    print(f"{'='*70}")
    print(f"Model:    facebook/mbart-large-50-many-to-many-mmt")
    print(f"Samples:  {len(predictions)}")
    print(f"BLEU:     {metrics['BLEU']:.2f}")
    print(f"METEOR:   {metrics['METEOR']:.2f}")
    print(f"chrF:     {metrics['chrF']:.2f}")
    print(f"ROUGE-L:  {metrics['ROUGE-L']:.4f}")
    print(f"Semantic: {metrics['Semantic']:.4f}")
    print(f"Emotion:  N/A (baseline has no emotion module)")
    print(f"{'='*70}\n")

    # Prepare results
    results = {
        'pair': pair,
        'model': 'facebook/mbart-large-50-many-to-many-mmt',
        'src_lang': src_mbart,
        'tgt_lang': tgt_mbart,
        'num_samples': len(predictions),
        'metrics': metrics,
        'timestamp': datetime.now().isoformat()
    }

    # Save results
    output_file = f"{OUTPUT_DIR}/mbart50_{pair}.json"
    with open(output_file, 'w', encoding='utf-8') as f:
        json.dump(results, f, indent=2, ensure_ascii=False)

    print(f"💾 Results saved: {output_file}\n")

    # Save sample translations
    sample_df = pd.DataFrame({
        'source': sources[:50],
        'reference': references[:50],
        'mbart50': predictions[:50]
    })
    sample_csv = f"{OUTPUT_DIR}/mbart50_samples_{pair}.csv"
    sample_df.to_csv(sample_csv, index=False, encoding='utf-8')
    print(f"💾 Samples saved: {sample_csv}\n")

    # Cleanup
    translator.cleanup()

    return results

# ============================================================================
# MAIN EXECUTION
# ============================================================================

def main():
    """Main evaluation pipeline"""

    print(f"\n{'='*80}")
    print("STARTING mBART-50 EVALUATION")
    print(f"{'='*80}\n")

    # Check if dataset exists
    if not os.path.exists(DATA_FILE):
        print(f"❌ Dataset not found: {DATA_FILE}")
        print("\nPlease update DATA_FILE path at the top of this script.")
        print("Example paths:")
        print("  - Kaggle: /kaggle/input/your-dataset/BHT25.csv")
        print("  - Colab:  /content/BHT25.csv")
        return

    # Evaluate all language pairs
    all_results = {}

    for pair in ['bn-hi', 'bn-te', 'hi-te']:
        try:
            results = evaluate_mbart50(pair, DATA_FILE)
            if results:
                all_results[pair] = results
        except Exception as e:
            print(f"❌ Error evaluating {pair}: {e}")
            import traceback
            traceback.print_exc()
            print()

        # Clear memory between pairs
        torch.cuda.empty_cache()
        gc.collect()

    # Save consolidated results
    if all_results:
        consolidated_file = f"{OUTPUT_DIR}/mbart50_all_pairs.json"
        with open(consolidated_file, 'w', encoding='utf-8') as f:
            json.dump(all_results, f, indent=2, ensure_ascii=False)

        print(f"\n💾 All results saved: {consolidated_file}")

        # Generate summary table
        generate_summary_table(all_results)

    print(f"\n{'='*80}")
    print("✅ EVALUATION COMPLETED!")
    print(f"{'='*80}\n")
    print(f"Results saved in: {OUTPUT_DIR}/")
    print(f"  - Individual results: mbart50_bn-hi.json, mbart50_bn-te.json, mbart50_hi-te.json")
    print(f"  - Consolidated: mbart50_all_pairs.json")
    print(f"  - Summary: mbart50_comparison_table.txt")
    print()

def generate_summary_table(all_results):
    """Generate comparison summary table"""

    print(f"\n{'='*90}")
    print("mBART-50 BASELINE RESULTS SUMMARY")
    print(f"{'='*90}\n")

    # Print table
    print(f"{'Pair':<10} {'Type':<15} {'BLEU':>10} {'METEOR':>10} {'chrF':>10} {'ROUGE-L':>10} {'Semantic':>10}")
    print("-" * 90)

    for pair in ['bn-hi', 'bn-te', 'hi-te']:
        if pair in all_results:
            r = all_results[pair]['metrics']
            pair_type = 'Within-family' if pair == 'bn-hi' else 'Cross-family'

            print(f"{pair:<10} {pair_type:<15} "
                  f"{r['BLEU']:>10.2f} "
                  f"{r['METEOR']:>10.2f} "
                  f"{r['chrF']:>10.2f} "
                  f"{r['ROUGE-L']:>10.4f} "
                  f"{r['Semantic']:>10.4f}")

    print(f"\n{'='*90}\n")

    # Save to file
    summary_file = f"{OUTPUT_DIR}/mbart50_comparison_table.txt"
    with open(summary_file, 'w') as f:
        f.write("mBART-50 BASELINE RESULTS\n")
        f.write("="*90 + "\n\n")
        f.write(f"Model: facebook/mbart-large-50-many-to-many-mmt (pre-trained, no fine-tuning)\n")
        f.write(f"Generated: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n\n")

        f.write(f"{'Pair':<10} {'Type':<15} {'BLEU':>10} {'METEOR':>10} {'chrF':>10} {'ROUGE-L':>10} {'Semantic':>10}\n")
        f.write("-" * 90 + "\n")

        for pair in ['bn-hi', 'bn-te', 'hi-te']:
            if pair in all_results:
                r = all_results[pair]['metrics']
                pair_type = 'Within-family' if pair == 'bn-hi' else 'Cross-family'

                f.write(f"{pair:<10} {pair_type:<15} "
                       f"{r['BLEU']:>10.2f} "
                       f"{r['METEOR']:>10.2f} "
                       f"{r['chrF']:>10.2f} "
                       f"{r['ROUGE-L']:>10.4f} "
                       f"{r['Semantic']:>10.4f}\n")

        f.write("\n" + "="*90 + "\n")

    print(f"💾 Summary table saved: {summary_file}\n")

if __name__ == "__main__":
    main()
```

# **M2M100 Baseline Evaluation**

```python
#!/usr/bin/env python3
"""
M2M100 Baseline Evaluation
M2M100 is specifically trained for 100 languages including Indic languages
MUCH BETTER than mBART-50 for Bengali/Hindi/Telugu!

NO TRAINING REQUIRED - Just evaluate pre-trained M2M100

Language codes:
- Bengali: 'bn' (ben)
- Hindi: 'hi' (hin)
- Telugu: 'te' (tel)
"""

import os
import torch
import pandas as pd
import numpy as np
from tqdm import tqdm
import json
from datetime import datetime
import gc

print("="*80)
print("M2M100 BASELINE EVALUATION")
print("Pre-trained Many-to-Many model for 100 languages")
print("="*80 + "\n")

# ============================================================================
# CONFIGURATION
# ============================================================================

DATA_FILE = '/content/ESA-NMT/BHT25_All.csv'
OUTPUT_DIR = './outputs'
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'

# M2M100 language codes (simple 2-letter codes)
M2M100_LANG_CODES = {
    'bn-hi': ('bn', 'hi'),  # Bengali to Hindi
    'bn-te': ('bn', 'te'),  # Bengali to Telugu
    'hi-te': ('hi', 'te')   # Hindi to Telugu
}

# Dataset column names
DATASET_COLUMNS = {
    'bn-hi': ('bn', 'hi'),
    'bn-te': ('bn', 'te'),
    'hi-te': ('hi', 'te')
}

TEST_LIMIT = 2000  # Number of test samples

print(f"Configuration:")
print(f"  Device: {DEVICE}")
print(f"  Dataset: {DATA_FILE}")
print(f"  Test samples: {TEST_LIMIT}")
print(f"  Output: {OUTPUT_DIR}\n")

os.makedirs(OUTPUT_DIR, exist_ok=True)

# ============================================================================
# LOAD DATASET
# ============================================================================

def load_data(csv_path, pair):
    """Load test split from BHT25 dataset"""
    print(f"\n{'='*70}")
    print(f"Loading dataset for {pair}")
    print(f"{'='*70}\n")

    df = pd.read_csv(csv_path)
    print(f"Total samples: {len(df)}")

    src_col, tgt_col = DATASET_COLUMNS[pair]

    if src_col not in df.columns or tgt_col not in df.columns:
        print(f"❌ ERROR: Columns '{src_col}' or '{tgt_col}' not found!")
        return None, None

    # Test split: last 15%
    train_size = int(len(df) * 0.70)
    val_size = int(len(df) * 0.15)
    test_start = train_size + val_size

    test_df = df[test_start:test_start+TEST_LIMIT]

    sources = test_df[src_col].astype(str).tolist()
    references = test_df[tgt_col].astype(str).tolist()

    print(f"✅ Test samples: {len(sources)}\n")
    print(f"Sample:")
    print(f"  SRC: {sources[0][:80]}...")
    print(f"  REF: {references[0][:80]}...\n")

    return sources, references

# ============================================================================
# M2M100 TRANSLATOR
# ============================================================================

class M2M100Translator:
    """M2M100 translator for Indic languages"""

    def __init__(self, src_lang, tgt_lang):
        """
        Initialize M2M100

        Args:
            src_lang: Source language ('bn', 'hi', 'te')
            tgt_lang: Target language ('bn', 'hi', 'te')
        """
        self.src_lang = src_lang
        self.tgt_lang = tgt_lang
        self.device = DEVICE

        print(f"Loading M2M100 model...")
        print(f"  Translation: {src_lang} → {tgt_lang}")

        from transformers import M2M100ForConditionalGeneration, M2M100Tokenizer

        # Use 418M model (good balance of speed/quality)
        # Alternative: "facebook/m2m100_1.2B" for better quality but slower
        model_name = "facebook/m2m100_418M"

        self.tokenizer = M2M100Tokenizer.from_pretrained(model_name)
        self.model = M2M100ForConditionalGeneration.from_pretrained(model_name)

        self.model = self.model.to(self.device)
        self.model.eval()

        # Set source language
        self.tokenizer.src_lang = src_lang

        print(f"✅ M2M100 loaded ({model_name})\n")

    def translate_batch(self, texts, batch_size=8):
        """Translate batch of texts"""
        translations = []

        print(f"Translating {len(texts)} samples...")

        with torch.no_grad():
            for i in tqdm(range(0, len(texts), batch_size), desc="M2M100"):
                batch_texts = texts[i:i+batch_size]

                try:
                    # Tokenize
                    encoded = self.tokenizer(
                        batch_texts,
                        return_tensors="pt",
                        padding=True,
                        truncation=True,
                        max_length=256
                    ).to(self.device)

                    # Generate with target language
                    generated = self.model.generate(
                        **encoded,
                        forced_bos_token_id=self.tokenizer.get_lang_id(self.tgt_lang),
                        max_length=256,
                        num_beams=5,
                        length_penalty=1.0,
                        early_stopping=True
                    )

                    # Decode
                    batch_translations = self.tokenizer.batch_decode(
                        generated,
                        skip_special_tokens=True
                    )

                    translations.extend(batch_translations)

                except Exception as e:
                    print(f"\n⚠️ Error in batch {i//batch_size}: {e}")
                    translations.extend([''] * len(batch_texts))

                if (i // batch_size) % 50 == 0:
                    torch.cuda.empty_cache()

        print(f"✅ Translation completed\n")
        return translations

    def cleanup(self):
        """Free GPU memory"""
        del self.model
        del self.tokenizer
        torch.cuda.empty_cache()
        gc.collect()

# ============================================================================
# METRICS
# ============================================================================

def calculate_metrics(predictions, references):
    """Calculate all metrics with proper type conversion for JSON"""
    print("Computing metrics...")

    metrics = {}

    # 1. BLEU
    print("  - BLEU...")
    from sacrebleu import corpus_bleu
    bleu = corpus_bleu(predictions, [references])
    metrics['BLEU'] = float(round(bleu.score, 2))  # Convert to Python float

    # 2. chrF
    print("  - chrF...")
    from sacrebleu.metrics import CHRF
    chrf = CHRF()
    chrf_score = chrf.corpus_score(predictions, [references])
    metrics['chrF'] = float(round(chrf_score.score, 2))

    # 3. METEOR
    print("  - METEOR...")
    try:
        import nltk
        nltk.download('wordnet', quiet=True)
        nltk.download('omw-1.4', quiet=True)
        nltk.download('punkt', quiet=True)

        from nltk.translate.meteor_score import meteor_score as calc_meteor

        meteor_scores = []
        for pred, ref in zip(predictions, references):
            try:
                pred_tokens = nltk.word_tokenize(pred) if pred else ['']
                ref_tokens = nltk.word_tokenize(ref) if ref else ['']
                score = calc_meteor([ref_tokens], pred_tokens)
                meteor_scores.append(score)
            except:
                meteor_scores.append(0.0)

        metrics['METEOR'] = float(round(np.mean(meteor_scores) * 100, 2))
    except:
        metrics['METEOR'] = 0.0

    # 4. ROUGE-L
    print("  - ROUGE-L...")
    try:
        from rouge_score import rouge_scorer
        rouge = rouge_scorer.RougeScorer(['rougeL'], use_stemmer=False)

        rouge_scores = []
        for pred, ref in zip(predictions, references):
            if pred and ref:
                score = rouge.score(ref, pred)
                rouge_scores.append(score['rougeL'].fmeasure)
            else:
                rouge_scores.append(0.0)

        metrics['ROUGE-L'] = float(round(np.mean(rouge_scores), 4))
    except:
        metrics['ROUGE-L'] = 0.0

    # 5. Semantic Similarity
    print("  - Semantic...")
    try:
        from sentence_transformers import SentenceTransformer

        semantic_model = SentenceTransformer('sentence-transformers/LaBSE').to(DEVICE)

        batch_size = 32
        all_cos_sims = []

        for i in range(0, len(predictions), batch_size):
            batch_ref = references[i:i+batch_size]
            batch_pred = predictions[i:i+batch_size]

            valid_pairs = [(r, p) for r, p in zip(batch_ref, batch_pred) if r and p]
            if not valid_pairs:
                continue

            batch_ref_valid, batch_pred_valid = zip(*valid_pairs)

            ref_emb = semantic_model.encode(list(batch_ref_valid), show_progress_bar=False)
            pred_emb = semantic_model.encode(list(batch_pred_valid), show_progress_bar=False)

            cos_sims = np.sum(ref_emb * pred_emb, axis=1) / (
                np.linalg.norm(ref_emb, axis=1) * np.linalg.norm(pred_emb, axis=1) + 1e-8
            )
            all_cos_sims.extend(cos_sims.tolist())  # Convert to list

        metrics['Semantic'] = float(round(np.mean(all_cos_sims), 4))

        del semantic_model
        torch.cuda.empty_cache()
    except:
        metrics['Semantic'] = 0.0

    print("✅ Metrics computed\n")
    return metrics

# ============================================================================
# EVALUATION
# ============================================================================

def evaluate_m2m100(pair, data_path):
    """Evaluate M2M100 on one language pair"""

    print(f"\n{'='*80}")
    print(f"EVALUATING: {pair.upper()}")
    print(f"{'='*80}\n")

    # Get language codes
    src_lang, tgt_lang = M2M100_LANG_CODES[pair]

    # Load data
    sources, references = load_data(data_path, pair)
    if sources is None:
        return None

    # Translate
    translator = M2M100Translator(src_lang, tgt_lang)
    predictions = translator.translate_batch(sources, batch_size=8)

    # Show samples
    print("Sample translations (first 5):")
    for i in range(min(5, len(predictions))):
        print(f"\n{i+1}.")
        print(f"  SRC:   {sources[i][:80]}...")
        print(f"  REF:   {references[i][:80]}...")
        print(f"  M2M:   {predictions[i][:80]}...")
    print()

    # Calculate metrics
    metrics = calculate_metrics(predictions, references)

    # Print results
    print(f"\n{'='*70}")
    print(f"M2M100 RESULTS: {pair}")
    print(f"{'='*70}")
    print(f"Model:    facebook/m2m100_418M")
    print(f"Samples:  {len(predictions)}")
    print(f"BLEU:     {metrics['BLEU']:.2f}")
    print(f"METEOR:   {metrics['METEOR']:.2f}")
    print(f"chrF:     {metrics['chrF']:.2f}")
    print(f"ROUGE-L:  {metrics['ROUGE-L']:.4f}")
    print(f"Semantic: {metrics['Semantic']:.4f}")
    print(f"Emotion:  N/A (baseline)")
    print(f"{'='*70}\n")

    # Prepare results
    results = {
        'pair': pair,
        'model': 'facebook/m2m100_418M',
        'src_lang': src_lang,
        'tgt_lang': tgt_lang,
        'num_samples': len(predictions),
        'metrics': metrics,
        'timestamp': datetime.now().isoformat()
    }

    # Save results
    output_file = f"{OUTPUT_DIR}/m2m100_{pair}.json"
    with open(output_file, 'w', encoding='utf-8') as f:
        json.dump(results, f, indent=2, ensure_ascii=False)

    print(f"💾 Results saved: {output_file}")

    # Save samples
    sample_df = pd.DataFrame({
        'source': sources[:50],
        'reference': references[:50],
        'm2m100': predictions[:50]
    })
    sample_csv = f"{OUTPUT_DIR}/m2m100_samples_{pair}.csv"
    sample_df.to_csv(sample_csv, index=False, encoding='utf-8')
    print(f"💾 Samples saved: {sample_csv}\n")

    # Cleanup
    translator.cleanup()

    return results

# ============================================================================
# MAIN
# ============================================================================

def main():
    """Main evaluation"""

    print(f"\n{'='*80}")
    print("STARTING M2M100 EVALUATION")
    print(f"{'='*80}\n")

    if not os.path.exists(DATA_FILE):
        print(f"❌ Dataset not found: {DATA_FILE}")
        return

    all_results = {}

    for pair in ['bn-hi', 'bn-te', 'hi-te']:
        try:
            results = evaluate_m2m100(pair, DATA_FILE)
            if results:
                all_results[pair] = results
        except Exception as e:
            print(f"❌ Error: {e}")
            import traceback
            traceback.print_exc()

        torch.cuda.empty_cache()
        gc.collect()

    # Save all results
    if all_results:
        consolidated = f"{OUTPUT_DIR}/m2m100_all_pairs.json"
        with open(consolidated, 'w', encoding='utf-8') as f:
            json.dump(all_results, f, indent=2, ensure_ascii=False)

        print(f"\n💾 All results: {consolidated}")

        # Summary table
        print(f"\n{'='*90}")
        print("M2M100 BASELINE RESULTS SUMMARY")
        print(f"{'='*90}\n")
        print(f"{'Pair':<10} {'Type':<15} {'BLEU':>10} {'METEOR':>10} {'chrF':>10} {'ROUGE-L':>10} {'Semantic':>10}")
        print("-" * 90)

        for pair in ['bn-hi', 'bn-te', 'hi-te']:
            if pair in all_results:
                r = all_results[pair]['metrics']
                pair_type = 'Within-family' if pair == 'bn-hi' else 'Cross-family'
                print(f"{pair:<10} {pair_type:<15} "
                      f"{r['BLEU']:>10.2f} {r['METEOR']:>10.2f} "
                      f"{r['chrF']:>10.2f} {r['ROUGE-L']:>10.4f} {r['Semantic']:>10.4f}")

        print(f"\n{'='*90}\n")

    print("✅ EVALUATION COMPLETED!\n")

if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
"""
IndicBART Baseline Evaluation
IndicBART is AI4Bharat's BART model for Indian languages
Supports: as, bn, gu, hi, kn, ml, mr, or, pa, ta, te

NO TRAINING REQUIRED - Just evaluate pre-trained IndicBART
"""

import os
import torch
import pandas as pd
import numpy as np
from tqdm import tqdm
import json
from datetime import datetime
import gc

print("="*80)
print("IndicBART BASELINE EVALUATION")
print("AI4Bharat's BART model for Indian languages")
print("="*80 + "\n")

# ============================================================================
# CONFIGURATION
# ============================================================================

DATA_FILE = '/content/ESA-NMT/BHT25_All.csv'
OUTPUT_DIR = './outputs'
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'

# IndicBART language codes
# IndicBART uses special format: <src_lang>_<tgt_lang> as task prefix
INDICBART_LANG_CODES = {
    'bn-hi': ('bengali', 'hindi', 'bn', 'hi'),
    'bn-te': ('bengali', 'telugu', 'bn', 'te'),
    'hi-te': ('hindi', 'telugu', 'hi', 'te')
}

TEST_LIMIT = 2000

print(f"Configuration:")
print(f"  Device: {DEVICE}")
print(f"  Dataset: {DATA_FILE}")
print(f"  Test samples: {TEST_LIMIT}")
print(f"  Output: {OUTPUT_DIR}\n")

os.makedirs(OUTPUT_DIR, exist_ok=True)

# ============================================================================
# CHECK INDICBART AVAILABILITY
# ============================================================================

def check_indicbart():
    """Check if IndicBART is available and which languages it supports"""
    print("Checking IndicBART model...")

    try:
        from transformers import MBartForConditionalGeneration, MBart50TokenizerFast

        # IndicBART model name
        model_name = "ai4bharat/IndicBART"

        print(f"Loading tokenizer from {model_name}...")
        tokenizer = MBart50TokenizerFast.from_pretrained(model_name)

        # Check supported languages
        if hasattr(tokenizer, 'lang_code_to_id'):
            supported = list(tokenizer.lang_code_to_id.keys())
            print(f"\n✅ IndicBART supports {len(supported)} languages")

            # Check our languages
            our_langs = ['bn_IN', 'hi_IN', 'te_IN']
            print(f"\nChecking our languages:")
            for lang in our_langs:
                if lang in supported:
                    print(f"  ✅ {lang} - SUPPORTED")
                else:
                    print(f"  ⚠️ {lang} - NOT FOUND")

            print(f"\nAll supported languages:")
            print(f"  {supported}")

        return True

    except Exception as e:
        print(f"❌ Error loading IndicBART: {e}")
        return False

# Check availability
indicbart_available = check_indicbart()
print()

if not indicbart_available:
    print("⚠️ IndicBART not available. Exiting.")
    exit(1)

# ============================================================================
# LOAD DATASET
# ============================================================================

def load_data(csv_path, pair):
    """Load test split from BHT25 dataset"""
    print(f"\n{'='*70}")
    print(f"Loading dataset for {pair}")
    print(f"{'='*70}\n")

    df = pd.read_csv(csv_path)
    print(f"Total samples: {len(df)}")

    _, _, src_col, tgt_col = INDICBART_LANG_CODES[pair]

    if src_col not in df.columns or tgt_col not in df.columns:
        print(f"❌ Columns not found!")
        return None, None

    # Test split (last 15%)
    train_size = int(len(df) * 0.70)
    val_size = int(len(df) * 0.15)
    test_start = train_size + val_size

    test_df = df[test_start:test_start+TEST_LIMIT]

    sources = test_df[src_col].astype(str).tolist()
    references = test_df[tgt_col].astype(str).tolist()

    print(f"✅ Test samples: {len(sources)}")
    print(f"Sample:")
    print(f"  SRC: {sources[0][:80]}...")
    print(f"  REF: {references[0][:80]}...\n")

    return sources, references

# ============================================================================
# INDICBART TRANSLATOR
# ============================================================================

class IndicBARTTranslator:
    """
    IndicBART translator for Indian languages
    """

    def __init__(self, src_lang, tgt_lang, src_name, tgt_name):
        """
        Initialize IndicBART

        Args:
            src_lang: Source lang code for dataset ('bn', 'hi', 'te')
            tgt_lang: Target lang code for dataset ('bn', 'hi', 'te')
            src_name: Source language name ('bengali', 'hindi', 'telugu')
            tgt_name: Target language name ('bengali', 'hindi', 'telugu')
        """
        self.src_lang = src_lang
        self.tgt_lang = tgt_lang
        self.src_name = src_name
        self.tgt_name = tgt_name
        self.device = DEVICE

        print(f"Loading IndicBART model...")
        print(f"  Translation: {src_name} → {tgt_name}")

        from transformers import MBartForConditionalGeneration, MBart50TokenizerFast

        model_name = "ai4bharat/IndicBART"

        try:
            self.tokenizer = MBart50TokenizerFast.from_pretrained(
                model_name,
                src_lang=f"{src_lang}_IN",
                tgt_lang=f"{tgt_lang}_IN"
            )
        except:
            # Try without lang parameters
            self.tokenizer = MBart50TokenizerFast.from_pretrained(model_name)

        self.model = MBartForConditionalGeneration.from_pretrained(model_name)

        self.model = self.model.to(self.device)
        self.model.eval()

        # Set source language
        try:
            self.tokenizer.src_lang = f"{src_lang}_IN"
        except:
            print(f"  ⚠️ Could not set src_lang to {src_lang}_IN")

        # Get target language ID
        try:
            self.tgt_lang_id = self.tokenizer.lang_code_to_id[f"{tgt_lang}_IN"]
        except:
            print(f"  ⚠️ Target language {tgt_lang}_IN not found")
            self.tgt_lang_id = None

        print(f"✅ IndicBART loaded\n")

    def translate_batch(self, texts, batch_size=8):
        """Translate batch of texts"""
        translations = []

        print(f"Translating {len(texts)} samples...")

        with torch.no_grad():
            for i in tqdm(range(0, len(texts), batch_size), desc="IndicBART"):
                batch_texts = texts[i:i+batch_size]

                try:
                    # Tokenize
                    encoded = self.tokenizer(
                        batch_texts,
                        return_tensors="pt",
                        padding=True,
                        truncation=True,
                        max_length=256
                    ).to(self.device)

                    # Generate
                    if self.tgt_lang_id is not None:
                        generated = self.model.generate(
                            **encoded,
                            forced_bos_token_id=self.tgt_lang_id,
                            max_length=256,
                            num_beams=5,
                            length_penalty=1.0,
                            early_stopping=True
                        )
                    else:
                        # Without forced target language
                        generated = self.model.generate(
                            **encoded,
                            max_length=256,
                            num_beams=5,
                            length_penalty=1.0,
                            early_stopping=True
                        )

                    # Decode
                    batch_translations = self.tokenizer.batch_decode(
                        generated,
                        skip_special_tokens=True
                    )

                    translations.extend(batch_translations)

                except Exception as e:
                    print(f"\n⚠️ Error in batch {i//batch_size}: {e}")
                    # Add empty translations as fallback
                    translations.extend([''] * len(batch_texts))

                # Memory cleanup
                if (i // batch_size) % 50 == 0:
                    torch.cuda.empty_cache()

        print(f"✅ Translation completed: {len(translations)} samples\n")
        return translations

    def cleanup(self):
        """Free GPU memory"""
        del self.model
        del self.tokenizer
        torch.cuda.empty_cache()
        gc.collect()

# ============================================================================
# METRICS (JSON-safe)
# ============================================================================

def calculate_metrics(predictions, references):
    """Calculate all metrics with proper type conversion"""
    print("Computing metrics...")

    metrics = {}

    # 1. BLEU
    print("  - BLEU...")
    from sacrebleu import corpus_bleu
    bleu = corpus_bleu(predictions, [references])
    metrics['BLEU'] = float(round(bleu.score, 2))

    # 2. chrF
    print("  - chrF...")
    from sacrebleu.metrics import CHRF
    chrf = CHRF()
    chrf_score = chrf.corpus_score(predictions, [references])
    metrics['chrF'] = float(round(chrf_score.score, 2))

    # 3. METEOR
    print("  - METEOR...")
    try:
        import nltk
        nltk.download('wordnet', quiet=True)
        nltk.download('omw-1.4', quiet=True)
        nltk.download('punkt', quiet=True)

        from nltk.translate.meteor_score import meteor_score as calc_meteor

        meteor_scores = []
        for pred, ref in zip(predictions, references):
            try:
                if not pred or not ref:
                    meteor_scores.append(0.0)
                    continue

                pred_tokens = nltk.word_tokenize(pred)
                ref_tokens = nltk.word_tokenize(ref)
                score = calc_meteor([ref_tokens], pred_tokens)
                meteor_scores.append(score)
            except:
                meteor_scores.append(0.0)

        metrics['METEOR'] = float(round(np.mean(meteor_scores) * 100, 2))
    except Exception as e:
        print(f"    ⚠️ METEOR calculation failed: {e}")
        metrics['METEOR'] = 0.0

    # 4. ROUGE-L
    print("  - ROUGE-L...")
    try:
        from rouge_score import rouge_scorer
        rouge = rouge_scorer.RougeScorer(['rougeL'], use_stemmer=False)

        rouge_scores = []
        for pred, ref in zip(predictions, references):
            if pred and ref:
                score = rouge.score(ref, pred)
                rouge_scores.append(score['rougeL'].fmeasure)
            else:
                rouge_scores.append(0.0)

        metrics['ROUGE-L'] = float(round(np.mean(rouge_scores), 4))
    except Exception as e:
        print(f"    ⚠️ ROUGE-L calculation failed: {e}")
        metrics['ROUGE-L'] = 0.0

    # 5. Semantic Similarity
    print("  - Semantic...")
    try:
        from sentence_transformers import SentenceTransformer

        semantic_model = SentenceTransformer('sentence-transformers/LaBSE').to(DEVICE)

        batch_size = 32
        all_cos_sims = []

        for i in range(0, len(predictions), batch_size):
            batch_ref = references[i:i+batch_size]
            batch_pred = predictions[i:i+batch_size]

            # Filter valid pairs
            valid_pairs = [(r, p) for r, p in zip(batch_ref, batch_pred) if r and p]
            if not valid_pairs:
                continue

            batch_ref_valid, batch_pred_valid = zip(*valid_pairs)

            # Encode
            ref_emb = semantic_model.encode(list(batch_ref_valid), show_progress_bar=False)
            pred_emb = semantic_model.encode(list(batch_pred_valid), show_progress_bar=False)

            # Cosine similarity
            cos_sims = np.sum(ref_emb * pred_emb, axis=1) / (
                np.linalg.norm(ref_emb, axis=1) * np.linalg.norm(pred_emb, axis=1) + 1e-8
            )
            all_cos_sims.extend(cos_sims.tolist())

        if all_cos_sims:
            metrics['Semantic'] = float(round(np.mean(all_cos_sims), 4))
        else:
            metrics['Semantic'] = 0.0

        del semantic_model
        torch.cuda.empty_cache()

    except Exception as e:
        print(f"    ⚠️ Semantic similarity calculation failed: {e}")
        metrics['Semantic'] = 0.0

    print("✅ Metrics computed\n")
    return metrics

# ============================================================================
# EVALUATION
# ============================================================================

def evaluate_indicbart(pair, data_path):
    """Evaluate IndicBART on one language pair"""

    print(f"\n{'='*80}")
    print(f"EVALUATING: {pair.upper()}")
    print(f"{'='*80}\n")

    # Get language codes
    src_name, tgt_name, src_code, tgt_code = INDICBART_LANG_CODES[pair]

    # Load data
    sources, references = load_data(data_path, pair)
    if sources is None:
        print(f"❌ Failed to load data for {pair}\n")
        return None

    # Initialize translator
    translator = IndicBARTTranslator(src_code, tgt_code, src_name, tgt_name)

    # Translate
    predictions = translator.translate_batch(sources, batch_size=8)

    # Show sample translations
    print("Sample translations (first 5):")
    for i in range(min(5, len(predictions))):
        print(f"\n{i+1}.")
        print(f"  SRC:       {sources[i][:80]}...")
        print(f"  REF:       {references[i][:80]}...")
        print(f"  IndicBART: {predictions[i][:80]}...")
    print()

    # Calculate metrics
    metrics = calculate_metrics(predictions, references)

    # Print results
    print(f"\n{'='*70}")
    print(f"IndicBART RESULTS: {pair}")
    print(f"{'='*70}")
    print(f"Model:    ai4bharat/IndicBART")
    print(f"Samples:  {len(predictions)}")
    print(f"BLEU:     {metrics['BLEU']:.2f}")
    print(f"METEOR:   {metrics['METEOR']:.2f}")
    print(f"chrF:     {metrics['chrF']:.2f}")
    print(f"ROUGE-L:  {metrics['ROUGE-L']:.4f}")
    print(f"Semantic: {metrics['Semantic']:.4f}")
    print(f"Emotion:  N/A (baseline has no emotion module)")
    print(f"{'='*70}\n")

    # Prepare results
    results = {
        'pair': pair,
        'model': 'ai4bharat/IndicBART',
        'src_lang': src_code,
        'tgt_lang': tgt_code,
        'num_samples': len(predictions),
        'metrics': metrics,
        'timestamp': datetime.now().isoformat()
    }

    # Save results
    output_file = f"{OUTPUT_DIR}/indicbart_{pair}.json"
    with open(output_file, 'w', encoding='utf-8') as f:
        json.dump(results, f, indent=2, ensure_ascii=False)

    print(f"💾 Results saved: {output_file}\n")

    # Save sample translations
    sample_df = pd.DataFrame({
        'source': sources[:50],
        'reference': references[:50],
        'indicbart': predictions[:50]
    })
    sample_csv = f"{OUTPUT_DIR}/indicbart_samples_{pair}.csv"
    sample_df.to_csv(sample_csv, index=False, encoding='utf-8')
    print(f"💾 Samples saved: {sample_csv}\n")

    # Cleanup
    translator.cleanup()

    return results

# ============================================================================
# MAIN EXECUTION
# ============================================================================

def main():
    """Main evaluation pipeline"""

    print(f"\n{'='*80}")
    print("STARTING INDICBART EVALUATION")
    print(f"{'='*80}\n")

    # Check if dataset exists
    if not os.path.exists(DATA_FILE):
        print(f"❌ Dataset not found: {DATA_FILE}")
        print("\nPlease update DATA_FILE path at the top of this script.")
        return

    # Evaluate all language pairs
    all_results = {}

    for pair in ['bn-hi', 'bn-te', 'hi-te']:
        try:
            results = evaluate_indicbart(pair, DATA_FILE)
            if results:
                all_results[pair] = results
        except Exception as e:
            print(f"❌ Error evaluating {pair}: {e}")
            import traceback
            traceback.print_exc()
            print()

        # Clear memory between pairs
        torch.cuda.empty_cache()
        gc.collect()

    # Save consolidated results
    if all_results:
        consolidated_file = f"{OUTPUT_DIR}/indicbart_all_pairs.json"
        with open(consolidated_file, 'w', encoding='utf-8') as f:
            json.dump(all_results, f, indent=2, ensure_ascii=False)

        print(f"\n💾 All results saved: {consolidated_file}")

        # Generate summary table
        print(f"\n{'='*90}")
        print("INDICBART BASELINE RESULTS SUMMARY")
        print(f"{'='*90}\n")
        print(f"{'Pair':<10} {'Type':<15} {'BLEU':>10} {'METEOR':>10} {'chrF':>10} {'ROUGE-L':>10} {'Semantic':>10}")
        print("-" * 90)

        for pair in ['bn-hi', 'bn-te', 'hi-te']:
            if pair in all_results:
                r = all_results[pair]['metrics']
                pair_type = 'Within-family' if pair == 'bn-hi' else 'Cross-family'

                print(f"{pair:<10} {pair_type:<15} "
                      f"{r['BLEU']:>10.2f} "
                      f"{r['METEOR']:>10.2f} "
                      f"{r['chrF']:>10.2f} "
                      f"{r['ROUGE-L']:>10.4f} "
                      f"{r['Semantic']:>10.4f}")

        print(f"\n{'='*90}\n")

        # Save summary
        summary_file = f"{OUTPUT_DIR}/indicbart_summary.txt"
        with open(summary_file, 'w') as f:
            f.write("IndicBART BASELINE RESULTS\n")
            f.write("="*90 + "\n\n")
            f.write(f"Model: ai4bharat/IndicBART\n")
            f.write(f"Generated: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n\n")

            f.write(f"{'Pair':<10} {'Type':<15} {'BLEU':>10} {'METEOR':>10} {'chrF':>10} {'ROUGE-L':>10} {'Semantic':>10}\n")
            f.write("-" * 90 + "\n")

            for pair in ['bn-hi', 'bn-te', 'hi-te']:
                if pair in all_results:
                    r = all_results[pair]['metrics']
                    pair_type = 'Within-family' if pair == 'bn-hi' else 'Cross-family'

                    f.write(f"{pair:<10} {pair_type:<15} "
                           f"{r['BLEU']:>10.2f} "
                           f"{r['METEOR']:>10.2f} "
                           f"{r['chrF']:>10.2f} "
                           f"{r['ROUGE-L']:>10.4f} "
                           f"{r['Semantic']:>10.4f}\n")

            f.write("\n" + "="*90 + "\n")

        print(f"💾 Summary saved: {summary_file}\n")

    print(f"{'='*80}")
    print("✅ EVALUATION COMPLETED!")
    print(f"{'='*80}\n")
    print(f"Results saved in: {OUTPUT_DIR}/")
    print(f"  - Individual: indicbart_bn-hi.json, indicbart_bn-te.json, indicbart_hi-te.json")
    print(f"  - Consolidated: indicbart_all_pairs.json")
    print(f"  - Summary: indicbart_summary.txt")
    print()

if __name__ == "__main__":
    main()
```
